```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

plt.style.use("ggplot")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)
```

```python
DATA_PATH = Path("data")

station = pd.read_csv("data/station_summary.csv")
inventory = pd.read_csv("data/inventory_items.csv")

print("Station Shape :", station.shape)
print("Inventory Shape :", inventory.shape)
```

```python
print("="*70)
print("STATION")
print("="*70)

display(station.head())

print(station.info())

print()

print("="*70)
print("INVENTORY")
print("="*70)

display(inventory.head())

print(inventory.info())
```

```python
print("Station Missing Values")

display(
    station.isna().sum().sort_values(ascending=False)
)

print()

print("Inventory Missing Values")

display(
    inventory.isna().sum().sort_values(ascending=False)
)
```

```python
print("Station duplicates")

print(station.duplicated().sum())

print()

print("Inventory duplicates")

print(inventory.duplicated().sum())
```
